Независимый Safety-агент для систем ИИ-поддержки студентов и подростков: детекция кризисных состояний и деструктивных воздействий на базе мультиагентной LLM-архитектуры с on-premise развертыванием


Что: Мы анализируем русскоязычные психологические данные (транскрипты сессий, клинические интервью, кризисные диалоги) для обучения мультиагентной LLM-системы детекции кризисных состояний у студентов и подростков.

Зачем: Для обучения мультиагентной архитектуры (агент-аналитик, агент-критик, агент-этик, Safety-агент) необходим собственный GPU-сервер за 5–8 млн рублей (2× RTX A6000 / L40S или 4× RTX 4090). Аренда аналогичной конфигурации стоит ~123 000 ₽/мес (1,5 млн ₽/год), но не обеспечивает соответствия 152-ФЗ — психологические данные не могут покидать вуз. Собственный сервер окупается за 4 года и становится активом кафедры для обучения, исследований и коммерциализации.

Почему это критически важно для позиций университета: Сеченовский университет — первый среди российских медицинских вузов в институциональном рейтинге THE (группа 1001–1200, 74-я позиция среди мировых медицинских университетов) и единственный российский вуз в топ-200 QS по «Медицине». В предметном рейтинге RAEX по направлению «Психология» университет занимает 6-е место, улучшив позицию. Однако для дальнейшего роста в этих рейтингах необходимы публикации в Q1 и внешнее финансирование — именно то, что даст предлагаемый проект. Без собственной вычислительной инфраструктуры мы не сможем конкурировать с НИУ ВШЭ, НГУ и зарубежными центрами за гранты РНФ и публикации в топ-журналах. Сервер — это условие сохранения лидерства в ИИ-психологии и укрепления позиций в мировых рейтингах.

Результаты: Публикации в журналах Q1 (Computers in Human Behavior Reports, Technology, Mind, and Behavior, Scientific Reports), заявка на грант РНФ (до 1,5 млн руб./год), Фонд Потанина (500 тыс. руб.) и РФРИТ (от 100 млн руб.).

Польза: Коммерциализация Safety-агента для кризисных центров и вузов (лицензия от 500 тыс. руб.) принесёт университету доход и укрепит позиции в области ИИ-психологии. Сервер — это не затраты, а инструмент для привлечения внешнего финансирования и роста в рейтингах.
